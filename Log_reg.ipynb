```python
import pandas as pd # importing main modules
import matplotlib.pyplot as plt 
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
from sklearn.metrics import roc_auc_score, roc_curve, accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
data=load_breast_cancer() #loading dataset
X=pd.DataFrame(data.data, columns=data.feature_names)#Feature columns
Y=pd.Series(data.target) #target 
X, x, Y, y = train_test_split(X, Y, test_size=0.2, random_state=42) # splitting into train and test 
```

```python
scaler= StandardScaler()
X = scaler.fit_transform(X)#standardizing
x=scaler.transform(x)
```

```python
model=LogisticRegression(max_iter=1000)#Applying Logistic Regression
model.fit(X, Y)
```

```python
pred=model.predict(x) #Predictions
pred
```

```python
acc=accuracy_score(y, pred) # calculating evalution metrics
prec=precision_score(y, pred)
rec=recall_score(y, pred)
f1=f1_score(y, pred)
auc_roc_scor=roc_auc_score(y, pred)
fpr, tpr, threshold=roc_curve(y, pred)
acc, prec, rec, f1, auc_roc_scor
```

```python
plt.figure() # plotting auc-roc curve
plt.plot(fpr, tpr, label=f'AUC-ROC:{auc_roc_scor}')
plt.plot([0,1],[0,1], linestyle='--')
plt.xlabel('False positive rate')
plt.ylabel('True positive rate')
plt.title('AUC-ROC curve')
plt.legend()
plt.show()
```
